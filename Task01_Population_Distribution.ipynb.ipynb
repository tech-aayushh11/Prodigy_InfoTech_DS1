## Task 01 - Distribution Visualization
**Dataset:** World Bank Population Data  
**Charts:** Region Distribution | Income Group Distribution

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Metadata_Country_API_SP.POP.TOTL_DS2_EN_csv_v2_327505.csv")
df.head()
```

```python
print(df.columns.tolist())
```

```python
plt.figure(figsize=(12, 6))
sns.countplot(y='Region', data=df, palette='Set2', order=df['Region'].value_counts().index)
plt.title('Number of Countries per Region')
plt.xlabel('Count')
plt.ylabel('Region')
plt.tight_layout()
plt.savefig('region_distribution.png')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(x='IncomeGroup', data=df, palette='viridis', order=df['IncomeGroup'].value_counts().index)
plt.title('Distribution of Countries by Income Group')
plt.xlabel('Income Group')
plt.ylabel('Count')
plt.xticks(rotation=15)
plt.tight_layout()
plt.savefig('income_distribution.png')
plt.show()
```
